```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
from pydantic import BaseModel, Field
```

```python
load_dotenv()
model = ChatGroq(model="openai/gpt-oss-20b",max_tokens=500)
```

```python
class EvaluationSchema(BaseModel):
    feedback: str =Field(description="Detailed feedback for the essay")
    score: int = Field(description="Score out of 10", ge=0, le=10)
    
```

```python
structured_model = model.with_structured_output(EvaluationSchema)
```

```python
essay = """Artificial Intelligence (AI) is rapidly becoming a major driver of economic growth, social development, and national strategy in India. Positioned as the 3rd most AI-competitive nation globally, India is leveraging AI to realize its long-term vision of Viksit Bharat 2047 (a developed nation by the centenary of its independence).

1. Key Economic and National Initiatives

• IndiaAI Mission: Backed by an allocation of over ₹10,300 crore, this flagship program builds critical domestic computing capacity, including the deployment of 38,000+ GPUs (subsidized for researchers and startups at ₹65/hour).
• Economic Contribution: AI is projected to add $1.7 trillion to the Indian economy by 2035, with technology sector revenues exceeding $280 billion and employing over 6 million people.
• Language and Data Inclusivity: Platforms like BHASHINI (supporting 20+ languages) and BharatGen AI (multilingual foundation model for 22 languages) help bridge the linguistic divide for millions of citizens.
• AIKosh and Digital ShramSetu provide open, high-quality public datasets and empower roughly 490 million informal-sector workers.

2. Sector-Wise Transformation

• Healthcare: AI-driven diagnostic tools are scaling access to remote areas, enabling early detection of tuberculosis, diabetic retinopathy, and cancer where specialist doctors are scarce. The Ayushman Bharat Digital Mission integrates AI-powered clinical decision support into public health infrastructure.
• Agriculture: Supporting nearly half of India's workforce, farming communities use smartphone imaging for instant crop-disease detection, soil analytics, precision irrigation, and localized weather forecasting via initiatives like Kisan AI.
• Education & Skills: Personalized learning platforms adapt to individual student paces, while automated administrative tools reduce teacher workloads. Institutions like the IITs and IISc have established dedicated AI research and degree programs.
• Governance and Smart Cities: AI streamlines public service delivery through automated fraud detection in welfare schemes, intelligent traffic management, and multilingual civic chatbots.

3. Core Challenges and Risks

• Job Displacement & Skill Mismatch: Rapid automation threatens routine assembly, manufacturing, and entry-level BPO/customer service roles, requiring aggressive workforce upskilling through frameworks like FutureSkills.
• The Digital Divide: Disparities in high-speed internet access and hardware between major metropolitan hubs and rural/tier-3 regions risk concentrating AI benefits among wealthier demographics.
• Data Privacy and Security: Scaling AI requires vast data collection, raising concerns over personal data protection, algorithmic bias, and AI-assisted cyber threats. The National Strategy for AI and emerging governance guidelines emphasize ethical principles centered on fairness, transparency, and accountability.

Conclusion

India’s journey with Artificial Intelligence is a balance between massive demographic potential and systemic infrastructure hurdles. If guided by inclusive public policy, robust digital safety nets, and local innovation, AI will transition from a high-tech novelty into the foundational infrastructure of a modern, self-reliant India."""
```

```python
prompt = f"Evaluate the language quality of the following essay and provide a feedback and assign a score out of 10 \n {essay}"
result = structured_model.invoke(prompt)
print(result)
```
